# Exploratory Data Analysis (EDA) 📊

## Objectives
- Understand feature distributions
- Identify outliers and anomalies
- Explore relationships between variables and credit risk
- Generate business insights
- Determine preprocessing requirements

```python
# Data manipulation
import pandas as pd
import numpy as np

# Visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)

# Plot style
plt.style.use('ggplot')
```

```python
# Load processed dataset

df = pd.read_csv('../data/processed/german_credit.csv')

df.head()
```

```python
# Dataset dimensions

df.shape
```

```python
# Dataset information

df.info()
```

```python
# Statistical summary

df.describe()
```

```python
# Numerical columns

num_cols = df.select_dtypes(
    include=['int64','float64']
).columns.tolist()

# Categorical columns

cat_cols = df.select_dtypes(
    include='object'
).columns.tolist()

print(num_cols)
print(cat_cols)
```

```python
# Class distribution

df['risk'].value_counts()
```

```python
# Percentage distribution

df['risk'].value_counts(normalize=True)*100
```

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x='risk'
)

plt.title('Credit Risk Distribution')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='age',
    bins=30,
    kde=True
)

plt.title('Age Distribution')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='credit_amount',
    bins=40,
    kde=True
)

plt.title('Credit Amount Distribution')

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='duration',
    bins=30,
    kde=True
)

plt.title('Loan Duration Distribution')

plt.show()
```

```python
numerical = [
    'age',
    'duration',
    'credit_amount'
]
```

```python
for col in numerical:

    plt.figure(figsize=(8,3))

    sns.boxplot(
        x=df[col]
    )

    plt.title(f'Boxplot of {col}')

    plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='risk',
    y='age'
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='risk',
    y='duration'
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='risk',
    y='credit_amount'
)

plt.show()
```

```python
num_cols = df.select_dtypes(
    include=['int64','float64']
).columns.tolist()

cat_cols = df.select_dtypes(
    include='object'
).columns.tolist()
```

```python
print(num_cols)
print(cat_cols)
```

```python
for col in cat_cols:

    plt.figure(figsize=(10,5))

    sns.countplot(
        data=df,
        x=col,
        hue='risk'
    )

    plt.xticks(rotation=45)

    plt.title(col)

    plt.show()
```

```python
corr_cols = [
    'duration',
    'credit_amount',
    'age',
    'installment_rate',
    'number_credits',
    'risk'
]

corr = df[corr_cols].corr()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Matrix')

plt.show()
```
