```python
import pandas as pd
```

```python
columns = [
    'status_checking_account',
    'duration',
    'credit_history',
    'purpose',
    'credit_amount',
    'savings_account',
    'employment_duration',
    'installment_rate',
    'personal_status_sex',
    'other_debtors',
    'present_residence',
    'property',
    'age',
    'other_installment_plans',
    'housing',
    'number_credits',
    'job',
    'people_liable',
    'telephone',
    'foreign_worker',
    'risk'
]
```

```python
df = pd.read_csv(
    '../data/raw/german.data',
    sep=' ',
    header=None,
    names=columns
)
```
