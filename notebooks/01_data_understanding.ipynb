# German Credit Risk Prediction - Data Understanding 📊

## Objective
The objective of this project is to predict whether a customer represents a good or bad credit risk using demographic and financial information.

## Dataset Source
UCI Machine Learning Repository - Statlog German Credit Dataset

## Tasks Completed
- Loaded dataset
- Assigned column names
- Examined dataset structure
- Converted target variable
- Checked missing values
- Analyzed target distribution
- Saved processed dataset

```python
# Import pandas for data manipulation and analysis
import pandas as pd
```

```python
# Since the original UCI dataset does not contain headers,
# we manually assign meaningful column names.
columns = [
    'status_checking_account',
    'duration',
    'credit_history',
    'purpose',
    'credit_amount',
    'savings_account',
    'employment_duration',
    'installment_rate',
    'personal_status_sex',
    'other_debtors',
    'present_residence',
    'property',
    'age',
    'other_installment_plans',
    'housing',
    'number_credits',
    'job',
    'people_liable',
    'telephone',
    'foreign_worker',
    'risk'
]
```

```python
# Load the German Credit dataset.
# The file is space-separated and does not contain headers.
df = pd.read_csv(
    '../data/raw/german.data',
    sep=' ',
    header=None,
    names=columns
)
```

```python
# Display the first five rows
# to understand the dataset structure.

df.head()
```

```python
# Determine the number of rows and columns.

df.shape
```

```python
# Check the distribution of credit risk classes.

df['risk'].value_counts()
```

```python
# Convert the target variable into meaningful labels.

df['risk'] = df['risk'].map({
    1: 'good',
    2: 'bad'
})
```

```python
# Display dataset information including:
# - Data types
# - Non-null counts
# - Memory usage

df.info()
```

```python
# Determine whether the dataset contains missing values.

df.isnull().sum()
```

```python
# Confirm successful conversion.

df['risk'].value_counts()
```

```python
# Import visualization libraries

import seaborn as sns
import matplotlib.pyplot as plt

# Plot class distribution.

sns.countplot(
    data=df,
    x='risk'
)

plt.title('Credit Risk Distribution')

plt.show()
```

```python
# Save the cleaned dataset
# for future notebooks.

df.to_csv(
    '../data/processed/german_credit.csv',
    index=False
)
```
